```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("clean_products.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
plt.figure(figsize=(10,5))
df['Category'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Product Categories')
plt.xlabel('Category')
plt.ylabel('Number of Products')
plt.xticks(rotation=45)
plt.show()
```

```python
Insight

The bar chart shows the Distribution of Products across different
categories. Clothing and appeal has the hishest number of products,
followed by Kitchen Appliances and Home & Kitchen. This indicates
that these categories dominate the product inventory.
```

```python
plt.figure(figsize=(10,5))
df['Brand'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Brands')
plt.xlabel('Brand')
plt.ylabel('Number of products')
plt.xticks(rotation=45)
plt.show()
```

```python
Insight

The analysis of brand distribution shows that no single brand
dominates the dataset. The top brands have very similar product
counts, indicating a balanced distribution of products across
multiple brands.
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Price'])
plt.title('Price Distribution')
plt.show()
```

```python
Insight

The boxplot shows that product prices are distributed
between 1 and 999. The median price is around 500,
indicating that most products are moderately priced.
There are no extreme outliers, suggesting a uniform
price distribution.
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Stock'], bins=20)
plt.title('Stock Distribution')
plt.xlabel('Stock')
plt.ylabel('Frequency')
plt.show()
```

```python
Insight

The stock distribution shows that most products
have a moderate stock level. Very few products
have  extremely low stock, and no products are
completely out of stock. This indicates good
investory management in the dataset.
```

```python
## Overall Analysis Insight

The analysis of the dataset shows that the
product distribution is well balanced across
categories and brands. Most products fall within
a moderate price, range, and inventory levels are
generally well maintained. There are only a few
low-stock products and no out-of stock items,
indicating efficient stock management.

```
